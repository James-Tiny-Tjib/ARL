```python
from unsloth import FastLanguageModel
import torch

# 1. Define your Hugging Face repo and local export path
# This is the exact name of your repo on Hugging Face
HF_MODEL_ID = "JamesResearch1216/llama-drone-adapter"
EXPORT_PATH = "./ARL/llama_drone_model"

# 2. Load directly from Hugging Face
# Unsloth will automatically download your adapter from HF, read its config,
# and download the correct base model to merge it with.
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = HF_MODEL_ID,
    max_seq_length = 4096,
    dtype = None,
    load_in_4bit = True,
)

print(f"Successfully loaded LoRA adapter from {HF_MODEL_ID} and base model.")

# 3. Export to GGUF for Ollama
print("Merging and exporting to GGUF format...")
model.save_pretrained_gguf(EXPORT_PATH, tokenizer, quantization_method="q4_k_m")

print(f"Recovery complete! GGUF model saved to: {EXPORT_PATH}")
```
